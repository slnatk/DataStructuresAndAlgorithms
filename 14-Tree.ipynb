```python
class Node:

    def __init__(self,value):

        self.value = value
        self.right = None
        self.left = None

# doubly linked list'te previous-next vardı        
```

```python
class BinarySearchTree:

    def __init__(self):

        self.root = None

    def insert(self, value):

        newNode = Node(value)

        if self.root is None:

            self.root = newNode
            return True
        
        tempNode = self.root # root'u değiştirmek istemiyoruz o yüzden böyle bir değere atadık
        
        while True:

            if newNode.value == tempNode.value:
                return False

            if newNode.value < tempNode.value:
                if tempNode.left is None: # eğer dal boşsa direkt ekle
                    tempNode.left = newNode
                    return True
                tempNode = tempNode.left # eğer doluysa da artık tempNode bir aşağıdaki. dal bir anda kök oldu ve artık ekleyeceği şeyi buranın altına ekleyecek.
            else:
                if tempNode.right is None:
                    tempNode.right = newNode
                    return True
                tempNode = tempNode.right # bu dağ sağ taraf için yani kökten büyükler için    
             
    def contains(self, value):

        tempNode = self.root

        while tempNode: # null olmadığı sürece demek bu.
            if value < tempNode.value:
                tempNode = tempNode.left # sola doğru ara
           
            elif value > tempNode.value:
                tempNode = tempNode.right
            else:
                return True
                return False # yoksa da false döndür

    def min0fNode(self, currentNode):
        while currentNode.left: # minimum aradığımız için, max olsa right derdik
            currentNode = currentNode.left
        return currentNode   

    def max0fNode(self, currentNode):
        while currentNode.right:
            currentNode = currentNode.right # null olmadığı sürece gittiği kadar git demek istiyoruz null gördüğü anda da durup son durduğu node u gösterecek
        return currentNode
```
