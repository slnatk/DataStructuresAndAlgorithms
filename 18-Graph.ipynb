```python
class Graph:

    def __init__(self):
        self.adjDict = {}

    def addVertex(self, vertex):
        if vertex not in self.adjDict.keys(): # yoksa ekle
            self.adjDict[vertex] = []
            return True
        return False

    def addEdge(self, v1,v2): # aralarında bağlantı var mı yok mu ona bakıyoruz.   
        if v1 in self.adjDict.keys() and v2 in self.adjDict.keys():
            self.adjDict[v1].append(v2)
            self.adjDict[v2].append(v1) # aralarında bağlantı kuruyoruz
            return True
        return False

    def removeEdge(self, v1,v2):
        if v1 in self.adjDict.keys() and v2 in self.adjDict.keys():
            try:
                self.adjDict[v1].remove(v2)
                self.adjDict[v2].remove(v1)
            except ValueError:
                pass # uygulama çökmesin diye böyle yapıyoruz
            return True
        return False

    def removeVertex(self, vertex):
        if vertex in self.adjDict.keys():
            for relatedVertex in self.adjDict[vertex]: # Sadece kendisini silmek yetmiyor diğer listelerde olup olmadığını kontrol edip diğerlerini de silmek lazım
                self.adjDict[relatedVertex].remove(vertex)
            del self.adjDict[vertex]
            return True
        return False

    def printGraph(self):
        for vertex in self.adjDict:
            print(vertex, "->", self.adjDict[vertex])
```

```python
myGraph = Graph()
myGraph.addVertex("IST")
myGraph.addVertex("AMS")
myGraph.addVertex("CDG")
myGraph.addVertex("JFK")

myGraph.addEdge("IST","AMS")
myGraph.addEdge("IST","CDG")
myGraph.addEdge("IST","JFK")
myGraph.addEdge("AMS","CDG")
myGraph.addEdge("AMS","JFK")
myGraph.addEdge("CDG","JFK")

myGraph.printGraph()
```

```python
myGraph.removeVertex("JFK")
myGraph.printGraph()
```

```python
myGraph.removeEdge("IST","CDG")
myGraph.removeEdge("CDG","JFK")
myGraph.printGraph()
```
