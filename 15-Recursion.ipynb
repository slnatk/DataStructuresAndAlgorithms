```python
def calculateFactorial(num):

    #edge case - exit condition, bunu yazmasak sonsuza kadar çalışırdı.
    if num == 0:
        return 1
    else:
        return num * calculateFactorial(num-1)

# 3 * calculateFactorial(2) -> 2
# 2 * calculateFactorial(1)-> 1
# 1 * calculateFactorial(0) = 1
```

```python
calculateFactorial(5)
```

```python
# Reverse String 
# write a function that reverses a string. The input string is given as an array of characters
# you must do this by modifying the input array in-place with O(1) extra memory. ekstra dizi oluşturmadan
```

```python
myArray = ["h","e","l","l","o"]
```

```python
myArray[0], myArray[-1] = myArray[-1], myArray[0]
```

```python
def reverseRecursive(s,start: int, end:int):
    if start > end:
        return
    s[start], s[end] = s[end], s[start]
    reverseRecursive(s,start+1, end-1)
```

```python
reverseRecursive(myArray,0,len(myArray)-1)
```
