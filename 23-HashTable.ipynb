```python
class HashTable:

    def __init__(self, size=13):
        self.dataMap = [None] * size # başlangıçta içinde hiçbir şey olmayan bir listemiz olacak

    def hashFunction(self,key):
        myHash = 0
        for letter in key:
            myHash = (myHash + ord(letter) * 19) % len(self.dataMap) # kapasite ayarı, yoksa çalışmaz
        return myHash

    def setItem(self, key, value): # anahtar - değer - kayıt
        index = self.hashFunction(key) # nereye kaydedeceğiz
        if self.dataMap[index] == None:
            self.dataMap[index] = [] # önce boş bir hale getiriyoruz -> [[]]
        self.dataMap[index].append([key,value])

    def getItem(self, key):
        index = self.hashFunction(key) # O(1) olmasını sağlayan burası
        if self.dataMap[index] is not None:
            for i in range(len(self.dataMap[index])): # listenin içinde gezme
                if self.dataMap[index][i][0] == key: # aradığımızı bulmuşuzdur
                    return self.dataMap[index][i][1]
            return None

    def getKeys(self):
        keys = []
        for i in range(len(self.dataMap)):
            if self.dataMap[i] is not None:
                for j in range(len(self.dataMap[i])): # listenin içinde liste old. için
                    keys.append(self.dataMap[i][j][0]) # key->0, value->1
        return keys

    def printTable(self):
        for index,value in enumerate(self.dataMap):
            print(index,"->",value)
                    
```

```python
myHashTable = HashTable()

myHashTable.setItem("Apple",100)
myHashTable.setItem("Banan",200)
myHashTable.setItem("Melon",300)

myHashTable.printTable()

```

```python
myHashTable.setItem("Watermelon",300)
myHashTable.setItem("Strawberry",100)
myHashTable.setItem("Lemon",150)

myHashTable.printTable()
```
